### Getting informations from tensor
1. Os tensores não estarem com o tipo certo de dado - para achar o tipo do dado, se usa `tensor.dtype`
2. Os tensores não terem o shape/formato correto - para descobri o shape de um tensor, se usa `tensor.shape`
3. Os tensores não estarem no dispositivo correto - para descobrir o dispositivo de um tensor, se usa `tensor.device`

```python
#Matrix Multiplication
Two main ways of performing multiplication in neural networks and deep learning:

Elemente-wise multiplication
Matrix Multiplication (dot product)

There are two main rules that performing matrix multiplication needs to satisfy:
1. The **inner dimensions** must match: 
* `(3,2) @ (3,2)` won't work
* `(2,3) @ (3,2)` will work
* `(3,2) @ (2,3)` will work
2. The resulting matrix has the shape of the **outer dimension**:
* `(2,3) @ (3,2) ` -> `(2,2)`


```

```python
torch.matmul(torch.rand(3,2),torch.rand(2,3))
```
